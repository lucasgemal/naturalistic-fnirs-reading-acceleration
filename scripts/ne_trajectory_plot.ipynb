# Neural Efficiency — Trajectory Plot
**Input: fnirs_neural_efficiency.xlsx**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Load
# Update this path to match your local data directory
PATH_IN  = r'../data/fnirs_neural_efficiency.xlsx'
# Update this path to match your local data directory
PATH_OUT = r'../results/neural_efficiency_trajectory.png'

df = pd.read_excel(PATH_IN)

# 2. Group labels for plot
df['group_label'] = df['group'].map({
    'acelerado'    : 'Accelerated',
    'nao_acelerado': 'Non-Accelerated'
})

# 3. Sort
df = df.sort_values('sessao_num')

print(f'Shape: {df.shape}')
print(f'ROIs: {df["roi"].unique().tolist()}')
print(f'Sessions: {sorted(df["sessao_num"].unique())}')
```

```python
# 4. Plot — one subplot per ROI
sns.set_theme(style='whitegrid')

fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

roi_config = [
    ('FRONTAL',  axes[0], 'Frontal ROI'),
    ('TEMPORAL', axes[1], 'Temporal ROI'),
]

palette = {
    'Accelerated'    : 'darkorange',
    'Non-Accelerated': 'steelblue'
}

for roi, ax, subtitle in roi_config:
    data = df[df['roi'] == roi]

    sns.lineplot(
        data=data,
        x='sessao_num',
        y='neural_efficiency',
        hue='group_label',
        style='group_label',
        markers=True,
        dashes=False,
        errorbar='se',
        palette=palette,
        linewidth=2,
        ax=ax
    )

    ax.axhline(0, linestyle='--', color='grey', linewidth=1)
    ax.set_title(subtitle, fontsize=13, fontweight='bold')
    ax.set_xlabel('Session', fontsize=11)
    ax.set_ylabel('Neural Efficiency (mean ± SEM)', fontsize=11)
    ax.set_xticks(range(1, 10))
    ax.legend(title='Group', fontsize=10)

fig.suptitle('Neural Efficiency Trajectory Across Sessions by Group and ROI',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(PATH_OUT, dpi=300, bbox_inches='tight')
print(f'-> Figure saved: {PATH_OUT}')
plt.show()
```
